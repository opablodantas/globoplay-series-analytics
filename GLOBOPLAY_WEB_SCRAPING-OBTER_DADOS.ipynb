### **Descrição do Objetivo do Projeto de Web Scraping no Globoplay**

O objetivo deste projeto é realizar um web scraping focado no catálogo de séries da plataforma Globoplay. Através desse processo, serão coletadas informações específicas sobre cada série, incluindo o nome, gênero, ano de lançamento e, possivelmente, uma breve descrição. O foco em séries permite uma análise detalhada da linha editorial e das estratégias de conteúdo que o Globoplay adota para atrair e fidelizar o público, além de entender quais tipos de gêneros e temas estão sendo priorizados.

Com a análise desses dados, será possível identificar padrões na seleção de conteúdo, como os gêneros mais prevalentes, as faixas etárias alvo, o tipo de narrativa predominante, e as eventuais parcerias ou licenças internacionais. Dessa forma, o projeto poderá fornecer insights sobre a estratégia de curadoria de conteúdo da plataforma, contribuindo para uma visão mais ampla da atuação da Globo no mercado de streaming e de entretenimento digital no Brasil.

#### Configurando o webdriver para extrair dados da página web

Configuração do Web Driver com Selenium e BeautifulSoup permite automatizar a navegação e extrair dados de páginas dinâmicas, possibilitando um scraping eficaz em sites complexos e com conteúdo carregado via JavaScript.

Essa é a URL que onde iremos obter os dados:

https://globoplay.globo.com/categorias/todas-as-series/

```python
# Pacotes que serão usados para fazer o web scraping no Globoplay

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException
from bs4 import BeautifulSoup
import time
import requests
import pandas as pd
```

```python
# Configurando o WebDriver 
driver = webdriver.Chrome()

# URL do Globoplay com as séries
driver.get("https://globoplay.globo.com/categorias/todas-as-series/")  
```

```python
# Criando um loop para clicar no botão "Mostrar mais" até ele desaparecer
while True:
    try:
        # Procura o botão "Mostrar mais" e clica nele
        botao_mostrar_mais = driver.find_element(By.XPATH, "//span[text()='Mostrar mais']")
        botao_mostrar_mais.click()
        
        # Aguarda alguns segundos para que os novos itens sejam carregados
        time.sleep(2)
    except NoSuchElementException:
        # Se o botão "Mostrar mais" não estiver mais presente, sai do loop
        break
```

```python
# Depois de carregar todos os itens, vamos extrair o conteúdo HTML da página
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
```

```python
# Extraindo os títulos das séries
titulo_serie = soup.find_all('h3', attrs={'class': "poster__caption"})
```

```python
# Vamos aproveitar e criar umas listas vazias que logo mais vai conter todo o conteúdo que pretendemos análisar

genero = []
ano_lancamento = []
classificacao = []
link = []
```

```python
# Agora vamos criar um loop para obter os títulos de todas as séries que há no globoplay
for title in titulo_serie:
    if title:
        nome = title.get_text(strip=True)
        titulos.append(nome)
```

```python
# Extraindo os links específicos de cada série
link_serie = soup.find_all('a', href = True) 

for serie in link_serie:
    href = serie.get('href')  # Obtém o valor de 'href'
    if href and href != '/':  # Ignora strings vazias ou '/'
        link_completo = f"https://globoplay.globo.com{href}"
        link.append(link_completo)
```

```python
# Exibe os títulos coletados
print(titulos)
```

```python
link = [
    links.strip("'[]") for links in link
]

print(link)
```

```python
# Lista para armazenar gêneros
generos_todas_series = []

# Iterar sobre cada link
for links in link:
    try:
        print(f"Processando link: {links}")
        response = requests.get(links)
        response.raise_for_status()  # Verifica se a resposta foi bem-sucedida
        soup = BeautifulSoup(response.content, 'html.parser')

        # Encontra todos os gêneros
        genres = soup.find_all('a', class_='genres__item-link')
        generos_serie = [genre.get_text(strip=True) for genre in genres]

        # Armazena os resultados
        generos_todas_series.append({
            "link": links,
            "generos": generos_serie
        })

    except Exception as e:
        print(f"Erro ao processar o link {links}: {e}")

# Exibindo os resultados
for serie in generos_todas_series:
    print(f"Série: {serie['link']}")
    print(f"Gêneros: {', '.join(serie['generos'])}")
```

```python

# Lista para armazenar classificações etárias
classificacoes_todas_series = []

for links in link:
    try:
        print(f"Processando link: {links}")
        response = requests.get(links)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, 'html.parser')

        # Encontra a classificação etária
        classificacao_elemento = soup.find('span', class_='content-rating__description')
        if classificacao_elemento:
            classificacao = classificacao_elemento.get_text(strip=True).replace('Classificação etária: ', '')
        else:
            classificacao = 'Não encontrada'

        # Armazena os resultados
        classificacoes_todas_series.append({
            "link": links,
            "classificacao_etaria": classificacao
        })

    except Exception as e:
        print(f"Erro ao processar o link {links}: {e}")
        classificacoes_todas_series.append({
            "link": links,
            "classificacao_etaria": 'Erro'
        })

```

```python

# Lista para armazenar os anos de lançamento
anos_todas_series = []

for links in link:
    try:
        print(f"Processando link: {links}")
        response = requests.get(links)
        response.raise_for_status()  # Verifica se a resposta foi bem-sucedida
        soup = BeautifulSoup(response.content, 'html.parser')

        # Encontra o ano de lançamento
        release_year = soup.find('li', class_='custom-description__item', id='custom-description-item-release-year')
        ano = release_year.get_text(strip=True) if release_year else "Ano não encontrado"

        # Armazena os resultados
        anos_todas_series.append({
            "link": links,
            "ano_lancamento": ano
        })

    except Exception as e:
        print(f"Erro ao processar o link {links}: {e}")

# Exibindo os resultados
for serie in anos_todas_series:
    print(f"Série: {serie['link']}")
    print(f"Ano de Lançamento: {serie['ano_lancamento']}")

# Convertendo para DataFrame
df_anos = pd.DataFrame(anos_todas_series)
print(df_anos)
```
